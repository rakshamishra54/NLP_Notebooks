### Using Spacy

```python
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
doc = nlp("A quick brown fox jumps over the lazy dog")
```

```python
doc.text
```

```python
print("Original Text: ", doc)
print("PoS Tagging Result:")
for token in doc:
    print(f"{token.text}: {token.pos_}")
```

####Using NLTK

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
text = "A quick brown fox jumps over the lazy dog."
words = word_tokenize(text)

pos_tags = pos_tag(words)
```

```python
print("Original text" , text)
print("POS tagging result :")
for word, pos_tag in pos_tags:
  print(f"{word}: {pos_tag}")
```

* NLTK’s default tagger is a lightweight model (averaged perceptron). It looks at a small window of nearby words, so with “quick brown fox” it can get confused and go with the most common tag for “brown”.
* spacy’s model is a neural network that uses richer context, so it’s usually more accurate on cases like this.

```python
for word in doc:
  print(word.text, "----->" ,word.pos_, word.tag_ , spacy.explain(word.tag_))
```
